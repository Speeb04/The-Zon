## My good friend Claude wrote this code
This cuck-coded piece of garbage turns this weird datetime format into week and holiday and whatever fatass is riding uber for 10 hours

```python
from pandas.tseries.holiday import USFederalHolidayCalendar

# 1. Convert strings to datetime
data_modify = data_has_tip.copy()

data_modify["tpep_pickup_datetime"] = pd.to_datetime(data_modify["tpep_pickup_datetime"], format="%Y-%m-%d %H:%M:%S")
data_modify["tpep_dropoff_datetime"] = pd.to_datetime(data_modify["tpep_dropoff_datetime"], format="%Y-%m-%d %H:%M:%S")

# 2. Trip duration in minutes
data_modify["trip_duration_min"] = (
    data_modify["tpep_dropoff_datetime"] - data_modify["tpep_pickup_datetime"]
).dt.total_seconds() / 60

# 3. Weekday / weekend (Mon=0 ... Sun=6)
data_modify["is_weekend"] = (data_modify["tpep_pickup_datetime"].dt.dayofweek >= 5).astype(int)

# 4. Holiday / not holiday (US federal holidays)
cal = USFederalHolidayCalendar()
holidays = cal.holidays(
    start=data_modify["tpep_pickup_datetime"].min(),
    end=data_modify["tpep_pickup_datetime"].max()
)
data_modify["is_holiday"] = data_modify["tpep_pickup_datetime"].dt.normalize().isin(holidays).astype(int)

# 5. Time of day based on pickup hour
hour = data_modify["tpep_pickup_datetime"].dt.hour
data_modify["time_of_day"] = np.select(
    [
        (hour >= 5) & (hour < 12),
        (hour >= 12) & (hour < 17),
        (hour >= 17) & (hour < 21),
    ],
    ["morning", "afternoon", "evening"],
    default="night",   # 21:00–04:59
)
```

### Encode zone to borough

```python
lookup = pd.read_csv("data/taxi_zone_lookup.csv")

id_to_value = dict(zip(lookup.iloc[:, 0], lookup.iloc[:, 1]))
id_to_value = lookup.set_index("LocationID")["Borough"].to_dict()
data_modify["pickup_borough"] = data_modify["PULocationID"].map(id_to_value)
data_modify["dropoff_borough"] = data_modify["DOLocationID"].map(id_to_value)
```
